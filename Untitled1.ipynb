```python
from scripts.spark_session import spark
spark.sparkContext.setLogLevel("ERROR")

from pyspark.sql.functions import col
from pyspark.sql import functions as F
```

```python
url1 = 'jdbc:mysql://localhost:3306/bronze'
```

```python
connect_properties = {
    'user' : 'root', 
    'password': 'macintosh',
    'driver': 'com.mysql.cj.jdbc.Driver'
}
```

```python
bronze_df = spark.read.jdbc(properties=connect_properties, url=url1, table='b_transaction')
bronze_df.printSchema()
```

```python
## renaming columns and appending a full name column
bronze_df.show(3)
```

```python
bronze_df.select(F.current_date()).show(2)
```

```python
bronze_df.select(F.current_timestamp()).show(2)
```

```python
## adding 5 days to current date
bronze_df = bronze_df.withColumn('curr_date', F.current_date())
bronze_df.show(2)
```

```python
bronze_df = bronze_df.select(col('curr_date'), F.date_add('curr_date', 30).alias('added days'))
```

```python
bronze_df.show(2)
```

```python
bronze_df = bronze_df.withColumn('date_year', F.year('curr_date'))
bronze_df.show(2)
```

```python
bronze_df = bronze_df.withColumn('day_of_week', F.dayofweek('curr_date'))
bronze_df = bronze_df.withColumn('is_weekend', F.when(col('day_of_week') == 1 | 7, True).otherwise(False))
bronze_df.show(2)
```
